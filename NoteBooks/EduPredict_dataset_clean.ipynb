```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
DATASET_PATH = 'raw_dataset.csv'
df = pd.read_csv(DATASET_PATH, index_col=False)
```

```python
def describe_dataset():
  for column in df.columns:
    nullValues = df[column].isnull()
    duplicated = df[column].duplicated().sum()
    print(f'[+] {column} Column Contain {nullValues.sum()} => {(nullValues.mean()):.2f}% Nulll Values | Duplicates : {duplicated}')

describe_dataset()
```

```python
def correct_scoreces_column():
  mask = df['Exam_Score'] > 100
  df.loc[mask, 'Exam_Score'] = 100

correct_scoreces_column()
```

```python
def fill_teacher_quality_column():
  nulls = df['Teacher_Quality'].isnull().values
  ids = np.array([id for id, boolean in enumerate(nulls) if boolean == True])
  for id in ids:
    row = df.iloc[id, 10:13]
    mask = (df['School_Type']  == row['School_Type']) & (df['Family_Income'] == row['Family_Income'])
    df.loc[id, 'Teacher_Quality'] = df.loc[mask, 'Teacher_Quality'].mode()[0]

fill_teacher_quality_column()
```

```python
describe_dataset()
```

```python
def fill_Parental_Education_Level_column():
  nulls = df['Parental_Education_Level'].isnull().values
  ids = np.array([id for id, boolean in enumerate(nulls) if boolean == True])
  for id in ids:
    row = df.iloc[id, 10:11]
    mask = df['Family_Income'] == row['Family_Income']
    df.loc[id, 'Parental_Education_Level'] = df.loc[mask, 'Parental_Education_Level'].mode()[0]

fill_Parental_Education_Level_column()
```

```python
describe_dataset()
```

```python
def fill_Distance_from_Home_column():
  nulls = df['Distance_from_Home'].isnull().values
  ids = np.array([id for id, boolean in enumerate(nulls) if boolean == True])
  for id in ids:
    row = df.iloc[id, 10:16]
    mask = (df['Family_Income'] == row['Family_Income']) & (df['School_Type'] == row['School_Type']) & (df['Learning_Disabilities'] == row['Learning_Disabilities'])
    df.loc[id, 'Distance_from_Home'] = df.loc[mask, 'Distance_from_Home'].mode()[0]

fill_Distance_from_Home_column()
```

```python
df.to_csv('cleaned_dataset.csv', index=False)
```
