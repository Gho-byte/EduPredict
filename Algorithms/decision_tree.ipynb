```python
class Question:
    def __init__(self, column, value):
        self.col = column
        self.value = value

    def match(self, element):
        value = element[self.col]
        if value.is_numeric():
            return value >= self.value
        return value == self.value

    def __repr__(self):
        condition = "=="
        if self.value.is_numeric():
            condition = ">="
        print(f'[+] Question is: {self.col} {condition} {self.value}')
```
